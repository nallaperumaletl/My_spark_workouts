```python
import pyspark
from pyspark.sql import SparkSession

spark = SparkSession.\
        builder.\
        appName("pyspark-notebook").\
        master("spark://spark-master:7077").\
        config("spark.executor.memory", "512m").\
        getOrCreate()
        # config("spark.executor.memory", "512m").\
        
```

```python
spark
```

```python
words = ("big", "data", "is", "very", "big", "area", "and","Very","INTERESTING","and", "VERY", "Funny")
```

```python
words_rdd = spark.sparkContext.parallelize(words)
```

```python
words_normalize = words_rdd.map(lambda x:x.lower())
```

```python
words_normalize.collect()
```

```python
mapped_words = words_normalize.map(lambda x: (x,1))
```

```python
words_agg = mapped_words.reduceByKey(lambda x,y : x+y)
```

```python
words_agg.collect()
```

```python
words_agg.sortBy(lambda x: x[1], False).collect()
```

# Or   ( in a single GO! )

```python
spark.sparkContext.parallelize(words).map(lambda x:x.lower()).map(lambda x:x.lower()).map(lambda x: (x,1)).reduceByKey(lambda x,y : x+y) \
.sortBy(lambda x: x[1], False).collect()
```
