```python
from pyspark.sql import SparkSession

spark = SparkSession \
    .builder \
    .appName("Read JSON data") \
    .master("spark://spark-master:7077") \
    .config("spark.executor.memory", "1g") \
    .config("spark.executor.cores", "2") \
    .config("spark.executor.instances", "3") \
    .getOrCreate()

spark
```
